```python
import re
```

```python
chat='your email should be like taimour@gmail.com'
chat1='123@gmail.com askdfklal lfskdfjlk dkfjslkd  3232@gmail.com'
```

```python
pattern = r'\b[a-zA-Z0-9._%+-]+@gmail\.com\b'
matches=re.findall(pattern, chat)
print(matches)
```

```python
patern1 = r'\b[a-zA-Z0-9._%+-]+@gmail\.com\b'
matches1=re.findall(patern1, chat1)
print(matches1)
```
