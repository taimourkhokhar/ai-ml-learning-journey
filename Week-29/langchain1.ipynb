```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

# Set your free Google API key
os.environ["GOOGLE_API_KEY"] = "YOUR_GOOGLE_API_KEY"  # Replace with your actual API key
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",  # or "gemini-1.5-flash"
    temperature=0.7
)
```

```python
response = llm.invoke("I want to open a bank account. What documents do I need to provide?")
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["cuisine"],
    template="Write a short and creative tweet about {cuisine} food."
)

formatted_prompt = prompt.format(cuisine="Italian")
print(formatted_prompt)
```

```python
from langchain.chains import LLMChain
llm_chain = LLMChain(llm=llm, prompt=prompt)
llm.chain.run(cuisine="Italian")
```
