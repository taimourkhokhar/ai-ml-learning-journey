```python
import math
def sigmoiod(x):
  return 1/(1+math.exp(-x))
```

```python
sigmoiod(100)
```

```python
sigmoiod(1)
```

```python
sigmoiod(-56)
```

```python
def tanh(x):
  return (math.exp(x)-math.exp(-x))/(math.exp(x)+math.exp(-x))
```

```python
tanh(-56)
```

```python
tanh(50)
```

```python
def relu(x):
  return max(0,x)
```

```python
relu(-7)
```

```python
relu(6)
```

```python
def leaky_relu(x):
  return max(0.1*x,x)
```

```python
leaky_relu(-100)
```

```python
leaky_relu(8)
```
