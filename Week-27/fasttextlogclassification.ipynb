```python
import pandas as pd
df=pd.read_csv("Ecommerce_data.csv")
df.head()
```

```python
df.label.value_counts
```

```python
df.dropna(inplace=True)
df.shape
```

```python
df['label'] = df['label'].replace("Clothing & Accessories", "Clothing_Accessories")
print(df['label'].unique())
```

```python
df['label']="__label__"+df['label'].astype(str)
```

```python
df.head()
```

```python
df['label_description']=df['label']+df['Text']
df.head()
```

```python
import re

text = "  VIKI's | Bookcase/Bookshelf (3-Shelf/Shelve, White) | ? . hi"
text = re.sub(r'[^\w\s\']',' ', text)
text = re.sub(' +', ' ', text)
text.strip().lower()
```

```python
def preprocess(text):
    text = re.sub(r'[^\w\s\']',' ', text)
    text = re.sub(' +', ' ', text)
    return text.strip().lower() 
```

```python
df['label_description'] = df['label_description'].map(preprocess)
df.head()
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(df, test_size=0.2)
```

```python

train.to_csv("ecommerce.train", columns=["label_description"], index=False, header=False)
test.to_csv("ecommerce.test", columns=["label_description"], index=False, header=False)
```

```python
import fasttext

model = fasttext.train_supervised(input="ecommerce.train")
model.test("ecommerce.test")
```

```python
import numpy as np

# Patch np.array to ignore breaking copy=False in NumPy 2.x
_orig_array = np.array
def _patched_array(obj, *args, **kwargs):
    kwargs.pop('copy', None)
    return _orig_array(obj, *args, **kwargs)

np.array = _patched_array

# Now run your prediction
model.predict("wintech assemble desktop pc cpu 500 gb sata hdd 4 gb ram intel c2d processor 3")
```
