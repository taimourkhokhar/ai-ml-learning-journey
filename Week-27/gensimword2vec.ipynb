```python
import gensim
from gensim.models import Word2Vec
```

```python
sentences = [
    ["i", "love", "dogs"],
    ["i", "love", "cats"],
    ["dogs", "are", "friendly"],
    ["cats", "are", "cute"],
    ["dogs", "play", "outside"],
    ["cats", "play", "inside"]
]
```

```python
model=Word2Vec(
  sentences=sentences,
  vector_size=100,
  window=5,
  min_count=1,
  workers=4
)
```

```python
word_vectors = model.wv["dogs"]
word_vectors
```

```python
model.wv.most_similar("dogs")
```
