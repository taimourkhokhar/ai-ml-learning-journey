```python
#practice3 build simple neural network
import math
def sigmoid(x):
  return 1/(1+math.exp(-x))

x1=2.0
x2=3.0

#neuron 1
n1_weight1=0.5
n1_weight2=-0.2
n1_bias=0.1
n1_pre_activation=(x1*n1_weight1)+(x2*n1_weight2)+n1_bias
n1_output=sigmoid(n1_pre_activation)

#neuron 2
n2_weight1 = 0.1
n2_weight2 = 0.8
n2_bias = -0.5

n2_pre_activation = (x1 * n2_weight1) + (x2 * n2_weight2) + n2_bias
n2_output = sigmoid(n2_pre_activation)

#layer 2 output neuron
out_weight1 = 0.4
out_weight2 = -0.6
out_bias = 0.2
out_pre_activation=(n1_output*out_weight1)+(n2_output*out_weight2)+out_bias
final_output=sigmoid(out_pre_activation)
```

```python
print("Neuron 1 output",round(n1_output,4))
print("neuron 2 output",round(n2_output,4))
print("Final network output",round(final_output,4))
```

```python
#practice 6 leaky relu
def leakRelu(x):
  if x>0:
    return x
  else:
    return 0.01*x
  
leakRelu(10)
```
