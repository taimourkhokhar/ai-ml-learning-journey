```python
def gradient_descent(w,a,gr):
  w_new=w-a*gr
  return w_new

gradient_descent(20,0.05,8)
```

```python
weight = 10
learning_rate = 0.1
gradients = [5, 4, 3, 2, 1]

for i,grad in enumerate(gradients,1):
  weight=weight-learning_rate*grad
  print(f"Iteration number is {i}")
  print(f"Gradient {grad} ")
  print(f"updated weights are {weight}")
```

```python
weight = 10
learning_rate = 0.1
gradients = [5, 4, 3, 2, 1]

for i, grad in enumerate(gradients, 1):
    weight = weight - learning_rate * grad
    print(f"Iteration {i}")
    print(f"Gradient: {grad}")
    print(f"Updated Weight: {weight:.1f}\n")
```

```python
weight = float(input("Enter an initial weight: "))
rate = float(input("Enter a learning rate: "))

gradients = [8, 6, 5, 3, 2]

for i, grad in enumerate(gradients, 1):
    weight = weight - rate * grad
    print(f"Iteration number is {i}")
    print(f"Updated weight is {weight:.2f}\n") 
```
