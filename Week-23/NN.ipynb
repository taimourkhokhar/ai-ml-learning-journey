```python
x_train_tensor=torch.from_numpy(x_train)
x_test_tensor=torch.from_numpy(x_test)
y_train_tensor=torch.from_numpy(y_train)
y_test_tensor=torch.from_numpy(y_test)
```

```python
#define model
import torch


# --- Class Definition ---
class MySimpleNN:

  def __init__(self, x) -> None:
    self.weights = torch.rand(
        x.shape[1], 1, dtype=torch.float64, requires_grad=True
    )
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, x):  # Fixed typo: forward
    z = torch.matmul(x, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
    # Fixed variable name: used y instead of y_train_tensor
    loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss


# --- Training Loop Execution Example ---
# Dummy data (ensure dtype is float64 to match weights)
X_train = torch.randn(100, 5, dtype=torch.float64)
y_train = torch.randint(0, 2, (100, 1), dtype=torch.float64)

model = MySimpleNN(X_train)
learning_rate = 0.1
epochs = 100

for epoch in range(epochs):
  # 1. Forward pass
  y_pred = model.forward(X_train)

  # 2. Compute loss
  loss = model.loss_function(y_pred, y_train)

  # 3. Backward pass
  loss.backward()

  # 4. Update parameters using gradient descent
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

    # Zero the gradients after updating
    model.weights.grad.zero_()
    model.bias.grad.zero_()

  if (epoch + 1) % 10 == 0:
    print(f'Epoch: {epoch+1}, Loss: {loss.item():.4f}')
```

```python
#create model
model = MySimpleNN(x_train_tensor)
learning_rate = 0.01

for epoch in range(epochs):
  y_pred = model.forward(x_train_tensor)

  loss = model.loss_function(y_pred, y_train_tensor)

  loss.backward()

  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

    model.weights.grad.zero_()
    model.bias.grad.zero_()

  print(f'Epoch: {epoch+1}, Loss: {loss.item():.4f}')
```
