```python
#autograd
x=torch.tensor(6.7)

y=torch.tensor(0.0)
w=torch.tensor(1.0)
b=torch.tensor(0.0)
```

```python
#Binay cross entropy loss for scalar
def binary_cross_entropy_loss(prediction,target):
  epsilon=1e-8
  prediction=torch.clamp(prediction,epsilon,1-epsilon)
  return -(target*torch.log(prediction)*torch.log(1-prediction))
```

```python
#forwad pass
z=w*x+b
y_pred=torch.sigmoid(z)

loss=binary_cross_entropy_loss(y_pred,y)
```

```python
loss
```

```python
#Derivates
dloss_dy_pred=(y_pred-y)/(y_pred*(1-y_pred))

dy_pred_dz=y_pred*(1-y_pred)
dz_dw=x
dz_db=1
dL_dw=dloss_dy_pred*dy_pred_dz*dz_dw
dL_db=dloss_dy_pred*dy_pred_dz*dz_db
```

```python
print(f"Manual Gradient of liss w.r.t weight (dw): {dL_dw}")
print(f"Manual Gradient of loss w.r.t weight (dw): {dL_db}")
```
